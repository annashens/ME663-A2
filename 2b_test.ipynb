```python
import sympy as sp

# Symbols
M, gamma = sp.symbols('M gamma', positive=True)
rho, a = sp.symbols('rho a', positive=True)
u = sp.symbols('u')

# Energy-flux scalar function g(M)
g = M*(sp.Rational(1,2)*M**2 + 1/(gamma-1))

# Unknown constant A in van Leer’s quartic ansatz
A = sp.symbols('A')

# Van Leer quartic ansatz for |M| ≤ 1
g_p = A*(M+1)**2*((gamma-1)*M + 2)**2
g_m = A*(M-1)**2*((gamma-1)*M - 2)**2

# Solve for A by enforcing g_p + g_m = g
eq = sp.expand(g_p + g_m - g)
A_sol = sp.solve(sp.Poly(eq, M).coeffs(), A)[0]

print("A =", A_sol)

# Substitute A back
g_p = sp.simplify(g_p.subs(A, A_sol))
g_m = sp.simplify(g_m.subs(A, A_sol))

# Convert back to fluxes
f1_p = rho*a/4*(M+1)**2
f1_m = rho*a/4*(M-1)**2

f3_p = rho*a**3 * g_p
f3_m = rho*a**3 * g_m

# Express in primitive variables (u = M*a)
f3_p_u = sp.simplify(f3_p.subs(M, u/a))
f1_p_u = sp.simplify(f1_p.subs(M, u/a))

ratio_p = sp.simplify(sp.factor(f3_p_u / f1_p_u))
print("ratio_p =", ratio_p)

# Final f3+ and f3- in primitive variables
f3_p_final = sp.simplify(f1_p_u * ratio_p)
print("f3_p(u) =", f3_p_final)
```
